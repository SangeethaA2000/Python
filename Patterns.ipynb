```python
# 1. RIGHT ANGLED TRIANGLE

*
* *
* * *
* * * *
* * * * *
```

```python
for i in range(5): # 0 1 2 3 4
    for j in range(i+1): # 1 2 3 4 5
        print("*", end=" ")  # end=" " - to print in same line
    print()  #after 1 iteration completes - to print in next line
```

```python
# 2. REVERSED RIGHT ANGLED TRIANGLE

* * * * *
* * * *
* * *
* * 
*
```

```python
num=int(input("enter the no."))
for i in range(num,0,-1):   # 5 - 5 4 3 2 1
    for j in range(i,0,-1): 
        print("*",end=" ")
    print()
```

```python
n=int(input("enter the no.:"))
for i in range(n,0,-1): # 6 - 6 5 4 3 2 1
    for j in range(1,i+1):  # 1, 6+1 = 1,7 = 1,2,3,4,5,6
                            # 1, 5+1 = 1,6 = 1,2,3,4,5
                            # 1, 4+1 = 1,5 = 1,2,3,4
                            # 1, 3+1 = 1,4 = 1,2,3
        print(j,end=" ")

    print()
```
